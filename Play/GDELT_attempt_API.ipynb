```python
import requests

# Define endpoint and parameters
url = "https://api.gdeltproject.org/api/v2/gkg/gkg"
params = {
    'query': 'S&P 500',
    'maxrecords': 100,
    'startdatetime': '20241001000000',  # Example: October 1, 2024, midnight
    'enddatetime': '20241031000000',    # Example: October 31, 2024, midnight
    'format': 'json'
}

# Send request
response = requests.get(url, params=params)

# Check for a successful response
if response.status_code == 200:
    data = response.json()
    # Verify the structure of the JSON response
    if 'articles' in data:
        # Process response
        for event in data['articles']:
            print(f"Title: {event['title']}")
            print(f"Date: {event['seendate']}")
            print(f"URL: {event['url']}")
            print("---------")
    else:
        print("No 'articles' key in the response data:", data)
else:
    print(f"Error {response.status_code}: {response.text}")
```
