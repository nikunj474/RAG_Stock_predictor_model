````python
Yes, I understand the possible function calls to the GDELT dataset API using the `gdeltdoc` library. Here are the main functions you can use:

1. **Article Search**: This function searches for news articles based on the specified filters.
    ```python
    articles = gd.article_search(f)
    ```

2. **Timeline Search**: This function searches for timelines based on the specified filters and timeline type.
    ```python
    timeline = gd.timeline_search('timelinevol', f)  # Volume timeline
    timeline_lang = gd.timeline_search('timelinelang', f)  # Language timeline
    timeline_tone = gd.timeline_search('timelinetone', f)  # Tone timeline
    ```

3. **Event Search**: This function searches for events based on the specified filters.
    ```python
    events = gd.event_search(f)
    ```

4. **Mention Search**: This function searches for mentions based on the specified filters.
    ```python
    mentions = gd.mention_search(f)
    ```

5. **Export Search**: This function exports search results based on the specified filters.
    ```python
    export = gd.export_search(f)
    ```

These functions allow you to interact with the GDELT dataset and retrieve various types of data based on your filters.
````
