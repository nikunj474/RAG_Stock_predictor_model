```python

gd2 = gdelt.gdelt(version=2)
news_data= gd2.Search(['2018 01 01','2024 01 01'], translation=True, table='gkg', output='pd')

# Print the DataFrame
print(news_data.head())
print(news_data.shape)
print(news_data.columns)
print(len(news_data))
```

```python
news_data.to_csv('news_data.csv', index=False)
```
