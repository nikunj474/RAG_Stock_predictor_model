```python
constituents = pd.read_csv('Data/stocks/constituents.csv')
print(constituents.head())
```

```python


# Create an empty dictionary to store the data for each ticker
ticker_data = {}

# Download the historical market data for the ticker
 # Create a list of ticker symbols
ticker_strings = constituents['Symbol'].tolist()

# Download the historical market data for all tickers
ticker_data = yf.download(ticker_strings, group_by='Ticker', start='2018-01-01', end='2024-01-01')
ticker_data = ticker_data.stack(level=0).rename_axis(['Date', 'Ticker']).reset_index(level=1)
```

```python
yf.Ticker('BRK.A').history(period='1d', interval='1m')
```

```python
#read ticker_data to csv
ticker_data.to_csv('ticker_data.csv')
```
